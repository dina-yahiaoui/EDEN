# Découverte de LangGraph

Objectif :

Comprendre les notions de :

- State
- Node
- Edge
- START
- END

```python
from typing_extensions import TypedDict
```

Le State représente la mémoire du graphe.

```python
class State(TypedDict):
    message: str
```

Création d'un Node.

Un Node est simplement une fonction.

```python
def dire_bonjour(state):
    return {
        "message": "Bonjour LangGraph !"
    }
```

```python
from langgraph.graph import StateGraph, START, END
```

```python
graph = StateGraph(State)
```

```python
graph.add_node("bonjour", dire_bonjour)
```

```python
graph.add_edge(START, "bonjour")
graph.add_edge("bonjour", END)
```

```python
app = graph.compile()
```

```python
result = app.invoke({})
print(result)
```
