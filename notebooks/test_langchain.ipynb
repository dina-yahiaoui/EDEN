```python
from langchain_core.prompts import ChatPromptTemplate
```

## PromptTemplate

Le PromptTemplate permet de créer un modèle de question avec des variables.

```python
prompt = ChatPromptTemplate.from_template(
    "Bonjour {nom}"
)
```

Ici, `{nom}` est une variable.
Elle sera remplacée automatiquement.

```python
message = prompt.invoke({"nom": "Dina"})
print(message)
```

Le PromptTemplate a remplacé automatiquement `{nom}` par "Dina".

Conclusion

Le PromptTemplate permet de créer facilement des prompts réutilisables.
